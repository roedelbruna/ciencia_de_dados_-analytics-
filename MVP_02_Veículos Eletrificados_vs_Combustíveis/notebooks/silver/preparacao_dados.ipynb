# Veículos Eletrificados vs Combustíveis

Nesta etapa foram selecionados os atributos relevantes para aa composição do racional e realizada agregações anuais para permitir a comparação entre a evolução dos veículos eletrificados e o consumo de combustíveis ao longo do período analisado.

## Transformação: seleção de colunas relevantes

Criação dos DataFrames com a seleção das colunas de interesse:
- **Gasolina e Etanol**: `ANO`, `CÓDIGO IBGE`, `MUNICÍPIO`, `VENDAS`
- **ABVE**: `ano`, `total_eletrificados`

Ao final, realiza uma validação da seleção exibindo as primeiras linhas de cada DataFrame transformado (`df_abve_silver`, `df_gasolina_silver` e `df_etanol_silver`) para conferência.

```python
import pandas as pd
```

```python
df_abve = pd.read_csv('../../data/abve_eletrificados_2020_2024.csv',
                      sep=';')

df_gasolina = pd.read_csv('../../data/anp_gasolina_c_2020_2024.csv', 
                          sep=';')

df_etanol = pd.read_csv('../../data/anp_etanol_hidratado_2020_2024.csv', 
                         sep=';')
```

```python
print(df_gasolina.columns.tolist())

print(df_etanol.columns.tolist())

print(df_abve.columns.tolist())
```

```python
df_gasolina.head()
df_etanol.head()

display(df_gasolina.head())
display(df_etanol.head())
```

```python
df_gasolina_silver  = df_gasolina[['ANO', 'CÓDIGO IBGE', 'MUNICÍPIO', 'VENDAS']].copy()

df_etanol_silver = df_etanol[['ANO','CÓDIGO IBGE', 'MUNICÍPIO', 'VENDAS']].copy()

df_abve_silver = df_abve[['ano', 'total_eletrificados']].copy()

df_abve_silver.head()
```

```python
df_gasolina_silver.head()
```

```python
df_etanol_silver.head()
```

```python
df_abve_anual = (df_abve_silver.groupby('ano', as_index=False)
                 ['total_eletrificados'].sum())

df_abve_anual                 
```

```python
df_gasolina_anual = (df_gasolina_silver.groupby('ANO', as_index=False)
                 ['VENDAS'].sum())

df_etanol_anual = (df_etanol_silver.groupby('ANO', as_index=False)
                 ['VENDAS'].sum())
```
