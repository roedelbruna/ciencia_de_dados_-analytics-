# Veículos Eletrificados vs Combustíveis

Neste notebook, os dados brutos são inspecionados e transformados: as colunas relevantes são selecionadas para compor os DataFrames que servirão como base para as análises e visualizações seguintes.

**Fontes de dados:**
- **ABVE** (Associação Brasileira de Veículos Elétricos): vendas de veículos eletrificados (2020–2024)
- **ANP** (Agência Nacional do Petróleo): produção de etanol hidratado e gasolina C (2020–2024)

## 1. Importação de bibliotecas

Importação da biblioteca `pandas` para manipulação dos dados tabulares.

## 2. Carga dos dados brutos

Carregamento dos arquivos CSV das fontes ABVE e ANP, no formato CSV com separador `;`, cobrindo o período de 2020 a 2024.

## 3. Inspeção das colunas

Exibição das colunas de cada DataFrame para identificar os campos disponíveis e planejar a seleção de variáveis relevantes.

## 4. Visualização inicial dos dados

Exibição das primeiras linhas dos DataFrames de gasolina e etanol para conferência inicial do conteúdo e dos formatos dos dados.

## 5. Transformação: seleção de colunas relevantes

Criação dos DataFrames com a seleção das colunas de interesse:
- **Gasolina e Etanol**: `ANO`, `CÓDIGO IBGE`, `MUNICÍPIO`, `VENDAS`
- **ABVE**: `ano`, `total_eletrificados`

Ao final, realiza uma validação da seleção exibindo as primeiras linhas de cada DataFrame transformado (`df_abve_silver`, `df_gasolina_silver` e `df_etanol_silver`) para conferência.

```python
import pandas as pd
```

```python
df_abve = pd.read_csv('../../data/abve_eletrificados_2020_2024.csv',
                      sep=';')

df_gasolina = pd.read_csv('../../data/anp_gasolina_c_2020_2024.csv', 
                          sep=';')

df_etanol = pd.read_csv('../../data/anp_etanol_hidratado_2020_2024.csv', 
                         sep=';')
```

```python
print(df_gasolina.columns.tolist())

print(df_etanol.columns.tolist())

print(df_abve.columns.tolist())
```

```python
df_gasolina.head()
df_etanol.head()

display(df_gasolina.head())
display(df_etanol.head())
```

```python
df_gasolina_silver  = df_gasolina[['ANO', 'CÓDIGO IBGE', 'MUNICÍPIO', 'VENDAS']].copy()

df_etanol_silver = df_etanol[['ANO','CÓDIGO IBGE', 'MUNICÍPIO', 'VENDAS']].copy()

df_abve_silver = df_abve[['ano', 'total_eletrificados']].copy()

df_abve_silver.head()
```

```python
df_gasolina_silver.head()
```

```python
df_etanol_silver.head()
```

## 6. Agregação anual

Agrupamento dos dados por ano para obter a evolução anual de cada base:
- **ABVE**: soma de `total_eletrificados` por `ano`
- **Gasolina e Etanol**: soma de `VENDAS` por `ANO`

Ao final, exibe os resultados de cada agregação para conferência.

```python
df_abve_anual = (df_abve_silver.groupby('ano', as_index=False)
                 ['total_eletrificados'].sum())

df_abve_anual                 
```

```python
df_gasolina_anual = (df_gasolina_silver.groupby('ANO', as_index=False)
                 ['VENDAS'].sum())

df_etanol_anual = (df_etanol_silver.groupby('ANO', as_index=False)
                 ['VENDAS'].sum())
```
